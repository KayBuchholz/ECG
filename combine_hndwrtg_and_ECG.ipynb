```python
#
# Cell 1: import libraries
#


import numpy as np
from pathlib import Path
import pandas as pd
import pickle
```

```python
#
# Cell 2: Specify the directory name for loading of data
#

from pathlib import Path

directory_path_ECG = Path("D:/ECG/XY_diff_class_0")
directory_path_handwriting = Path("D:/ECG/handwriting")
```

```python
#
# Cell 3: load ECG X-data and handwriting X-data 
#

#loaded_arrays_ECG = np.load(directory_path_ECG/'X_np_arrays.npz',allow_pickle=True)
#loaded_arrays_ECG = np.load(directory_path_ECG/'X_np_arrays_2.npz',allow_pickle=True)
loaded_arrays_ECG = np.load(directory_path_ECG/'X_np_arrays_3.npz',allow_pickle=True)

# chose the file to restore:
X_train_ECG = loaded_arrays_ECG['a']
X_val_ECG = loaded_arrays_ECG['b']
X_test_ECG = loaded_arrays_ECG['c']

print("X_train_ECG shape:", X_train_ECG.shape)
print("X_val_ECG:", X_val_ECG.shape)
print("X_test_ECG:", X_test_ECG.shape)

#loaded_arrays_handwriting = np.load(directory_path_handwriting/'X_hwrtng_np_arrays.npz',allow_pickle=True)
#loaded_arrays_handwriting = np.load(directory_path_handwriting/'X_hwrtng_np_arrays_2.npz',allow_pickle=True)
loaded_arrays_handwriting = np.load(directory_path_handwriting/'X_hwrtng_np_arrays_3.npz',allow_pickle=True)

# chose the file to restore:
X_train_handwriting = loaded_arrays_handwriting['a']
X_val_handwriting = loaded_arrays_handwriting['b']
X_test_handwriting = loaded_arrays_handwriting['c']

print("X_train_handwriting shape:", X_train_handwriting.shape)
print("X_val_handwriting:", X_val_handwriting.shape)
print("X_test_handwriting:", X_test_handwriting.shape)
```

```python
#
# Cell 4: Putt ECG X-data and handwriting X-data into ddictionaries for later processing 
#

X_ECG={'X_train' : X_train_ECG,
       'X_val' : X_val_ECG,
       'X_test' : X_test_ECG}

X_handwriting={'X_train' : X_train_handwriting,
       'X_val' : X_val_handwriting,
       'X_test' : X_test_handwriting}
```

```python
#
# Cell 5: load ECG y-data and handwriting y-data   
#

#with open(directory_path_ECG/'Y_data.pkl', 'rb') as f:
#    y_data_ECG = pickle.load(f) # deserialize using load()
#f.close()  

#with open(directory_path_handwriting/'Y_hwrtng_data.pkl', 'rb') as f:
#    y_data_handwriting = pickle.load(f) # deserialize using load()
#f.close()

#==================

#with open(directory_path_ECG/'Y_data_2.pkl', 'rb') as f:
#    y_data_ECG = pickle.load(f) # deserialize using load()
#f.close()  

#with open(directory_path_handwriting/'Y_hwrtng_data_2.pkl', 'rb') as f:
#    y_data_handwriting = pickle.load(f) # deserialize using load()
#f.close()

#=====================

with open(directory_path_ECG/'Y_data_3.pkl', 'rb') as f:
    y_data_ECG = pickle.load(f) # deserialize using load()
f.close()  

with open(directory_path_handwriting/'Y_hwrtng_data_3.pkl', 'rb') as f:
    y_data_handwriting = pickle.load(f) # deserialize using load()
f.close()
```

```python
#
# Cell 5: for analytical purpose only 
#

y_data_ECG
```

```python
#
# Cell 6: combines operations of cell 6 and cell 7 for train_df, val_df and test_df   
#


y_data_ECG_modification_1 = {name: df['codes_2'].copy()
                             for name, df in y_data_ECG.items()}


#individual_dfs = {
#    name: Y_label_len_one_short_df[Y_label_len_one_short_df['codes'] == code].copy()
#    for name, code in dfs.items()
#}

y_data_ECG_modification_1['val_df']
```

```python
#
# Cell 7: combines operations of cell 9 and cell 10 for train_df, val_df and test_df   
#


y_data_handwriting_modification_1 = {name: df['sentence'].copy()
                             for name, df in y_data_handwriting.items()}

y_data_handwriting_modification_1['train_df']
```

```python
#
# Cell 8: combines operations of cell 10 and cell 11 for train_df, val_df and test_df      
#

y_data_combined = {key:pd.concat([y_data_handwriting_modification_1[key], 
                                                      y_data_ECG_modification_1[key]], axis=0, ignore_index=True)
                             for key in y_data_handwriting_modification_1.keys()}

y_data_combined['val_df']

```

```python
#
# Cell 9: concatenate the numpy arrays of the dictionary X_handwriting and X_ECG
#

X_data_modification_1 = {key:np.concatenate([X_handwriting[key],
                                        X_ECG[key]], axis=0)
                             for key in X_handwriting.keys()}

X_data_modification_1['X_train'].shape
```

```python
#
# Cell 9a: for analytical purpose
#

print (X_data_modification_1['X_train'] [10762,0,0])
print (X_data_modification_1['X_train'] [9045,0,0])
print (X_data_modification_1['X_train'] [1105,0,0])
print (X_data_modification_1['X_train'] [5111,0,0])

#10762              all_other
#9045               all_other
#1105        fulfilling Dylks
#5111                      SR
```

```python
#
# Cell 10: shuffle 'y_combined' for y_train, y_val, y_test
#


y_data_combined_shuffled = {name: df.sample(frac=1).copy()
                             for name, df in y_data_combined.items()}

y_data_combined_shuffled['train_df'][0:50]
```

```python
#
# Cell 10: shuffle X_train, X_val_X_test of dictionary X_data according y_data_combined_shuffled
#

key_map = {
    "train_df": "X_train",
    "val_df": "X_val",
    "test_df": "X_test"
}

X_shuffled={}

for y_key, x_key in key_map.items():
    X_shuffled[x_key]= X_data_modification_1[x_key][y_data_combined_shuffled[y_key].index.tolist(), :, :]
X_shuffled['X_train'].shape
```
