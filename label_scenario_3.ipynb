```python
#
# Cell 3: Definition of functions
#

def load_raw_data(df, sampling_rate, path):
    if sampling_rate == 100:
        data = [wfdb.rdsamp(path+f) for f in df.filename_lr]
    else:
        data = [wfdb.rdsamp(path+f) for f in df.filename_hr]
    data = np.array([signal for signal, meta in data])
    return data

def aggregate_diagnostic(y_dic):
    tmp = []
    for key in y_dic.keys():
        if key in agg_df.index:
            tmp.append(agg_df.loc[key].diagnostic_class)
    return list(set(tmp))

def aggregate_form_rhythm(y_dic): #BuK
    tmp = []
    for key in y_dic.keys():
        if key in agg_df.index:
            tmp.append(agg_df.loc[key].form_rhythm)
    return list(set(tmp))

def aggregate_form(y_dic): #BuK
    tmp = []
    for key in y_dic.keys():
        if key in agg_df_form.index:
            tmp.append(agg_df.loc[key].form)
    return list(set(tmp))
    
```

```python
#
# Cell 10: in all combinations of labels with "SR" remove "SR" 
# clean list of labels is now stored in the column "codes"
#

Y_copy["codes"] = Y_copy["label"].apply(
    lambda lst: ["SR"] if len(lst) == 1 and lst[0] == "SR"
    else sorted([c for c in lst if c != "SR"])
)
Y_copy
```

```python
#
#Cell 11: determines length of lists "codes" and add the sesult to the column "len"
#

Y_copy["len"] = Y_copy["codes"].apply(
    lambda lst: len(lst)
)
Y_copy
```

```python
#
# Cell 12: filters all instances with length zero (only for control purposes) 
# Note: these instances are ECGs with neither form nor rhythm statments but tagged with a diagnostic label.
# Very few instance are labelled with "NORM". These ECGs could have been added to the SR class but this ...  
#

Y_copy = Y_copy[Y_copy["len"]>0].copy()
Y_copy
```

```python
#
# Cell 13: filters all instances with length one. 
# These instances contain all instances with one single Form or rhythm statement.    
#

Y_copy_len_one_df = Y_copy[Y_copy["len"]==1].copy()
Y_copy_len_one_df
```

```python
#
# Cell 14: substract Y_copy_len_one_df Y_copy to get Y_all_other
#

Y_all_other = Y_copy[~Y_copy.index.isin(Y_copy_len_one_df.index)].copy()
Y_all_other
```

```python
#
# Cell 15: convert list element of column "codes" into objects
#

Y_copy_len_one_df['codes'] = Y_copy_len_one_df['codes'].apply(lambda x: " ".join(x))
Y_copy_len_one_df
```

```python
#
# Cell 16: filters rows from Y_copy by index of agg_df_form 
# that is, filter all form ECGs
#

Y_form_df=Y_copy_len_one_df[Y_copy_len_one_df['codes'].isin(agg_df_form.index)].copy()
Y_form_df
```

```python
#
# Cell 17: filters rows from Y_copy by index of agg_df_rhythm 
# that is, filter all rhythm ECGs
#

Y_rhythm_df=Y_copy_len_one_df[Y_copy_len_one_df['codes'].isin(agg_df_rhythm.index)].copy()
Y_rhythm_df
```

```python
#
# Cell 18: selects all ECGS with "SR" label resp. code  
#

Y_SR_df = Y_rhythm_df[Y_rhythm_df['codes'] == 'SR'].copy()
Y_SR_df
```

```python
#
# Cell 19: selects all Non SR rhythm labelled ECGs  
#

Y_rhythm_no_SR_df = Y_rhythm_df[~Y_rhythm_df.index.isin(Y_SR_df.index)].copy()
Y_rhythm_no_SR_df
```

```python
#
# Cell 20: the instances tagged with a value 10 in the columns "strat_forld" are revied carefully 
# by a cardiologist and therefore these instances are intended for the test course.
# NOTE! these cell serves only for information about the length resp. number of rows of the corresponding df 
#

#Y_rhythm_test = Y_rhythm_no_SR_df[Y_rhythm_no_SR_df.strat_fold == 10]
#Y_rhythm_test

#Y_all_other_test = Y_all_other[Y_all_other.strat_fold == 10]
#Y_all_other_test

Y_SR_test = Y_SR_df[Y_SR_df.strat_fold == 10]
Y_SR_test
```

```python
#
# Cell 21:  assign a unique code to all ECGs of one class (form, rhytm, all other)    
#

df_list= [[Y_form_df, 'form'], [Y_rhythm_no_SR_df, 'rhythm'], [Y_all_other, 'all_other']]

df_list
for item in df_list:
    #item[0]['codes_recover'] = item[0]['codes']
    item[0]['codes'] = item[1]

Y_rhythm_df
```

```python
#
# Cell 22: splits the test instances from the different classes by the strat_fold value 10.
#

dfs = {
    'form': Y_form_df,
    'rhythm_no_SR': Y_rhythm_no_SR_df,
    'all_other': Y_all_other,
    'SR': Y_SR_df
}

test_dfs = {name: df[df['strat_fold'] == 10].copy() for name, df in dfs.items()}
print('len test_dfs[SR]: ', len(test_dfs['SR']))
print('len test_dfs[form]: ', len(test_dfs['form']))
print('len test_dfs[rhythm_no_SR]: ', len(test_dfs['rhythm_no_SR']))
print('len test_dfs[all_other]: ', len(test_dfs['all_other']))
```

```python
test_dfs['all_other']
```

```python
#
# Cell 23: substracts for all classes the corresponding test instances from the sets.
# Generates the dictionary "residue_dfs" with all the residue df of the individual classes  
#

dfs = {
    'form_res': [Y_form_df, test_dfs['form']],
    'rhythm_no_SR_res': [Y_rhythm_no_SR_df, test_dfs['rhythm_no_SR']],
    'all_other_res': [Y_all_other, test_dfs['all_other']],
    'SR_res': [Y_SR_df, test_dfs['SR']]
}


residue_dfs = {name: df[0][~df[0].index.isin(df[1].index)].copy() for name, df in dfs.items()}

residue_dfs['SR_res']


print('len residue_dfs[SR_res]: ', len(residue_dfs['SR_res']))
print('len residue_dfs[form_res]: ', len(residue_dfs['form_res']))
print('len residue_dfs[rhythm_no_SR_res]: ', len(residue_dfs['rhythm_no_SR_res']))
print('len residue_dfs[all_other_res]: ', len(residue_dfs['all_other_res']))
```

```python
residue_dfs['all_other_res']
```

```python
#
# Cell 24: splits the df in the dictionary "residue_dfs" in training and validation sets.
#

dfs = {
    'form_training': residue_dfs['form_res'],
    'rhythm_training': residue_dfs['rhythm_no_SR_res'],
    'all_other_training': residue_dfs['all_other_res'],
    'SR_training': residue_dfs['SR_res']
}

training_dfs = {
    name: df.sample(frac=0.75, random_state=7).copy()
    for name, df in dfs.items()
}

print('len training_dfs[SR_training]: ', len(training_dfs['SR_training']))
print('len training_dfs[form_training]: ', len(training_dfs['form_training']))
print('len training_dfs[rhythm_training]: ', len(training_dfs['rhythm_training']))
print('len training_dfs[all_other_training]: ', len(training_dfs['all_other_training']), '\n')


validation_dfs = {
    name.replace('_training', '_validation'): df.drop(training_dfs[name].index).copy()
    for name, df in dfs.items()
}

print('len validation_dfs[SR_validation]: ', len(validation_dfs['SR_validation']))
print('len validation_dfs[form_validation]: ', len(validation_dfs['form_validation']))
print('len validation_dfs[rhythm_validation]: ', len(validation_dfs['rhythm_validation']))
print('len validation_dfs[all_other_validation]: ', len(validation_dfs['all_other_validation']))

display(training_dfs['SR_training'][0:5]) 
display(validation_dfs['SR_validation'][0:5])
```

```python
#
# Cell 25: count the number of instances splits the df in the dictionary "residue_dfs" in training and validation sets.
#

dfs_all= {1: test_dfs,
          2: training_dfs,
          3: validation_dfs
         }

summation=sum(df.shape[0] for df in dfs_all.values() for df in df.values())
print('Total sum of all instances in the different sets: ',summation)
```

```python
#
# Cell 26:
#

train_df = pd.concat(training_dfs.values(), axis=0).sample(frac=1, random_state=7)
val_df   = pd.concat(validation_dfs.values(), axis=0).sample(frac=1, random_state=7)
test_df  = pd.concat(test_dfs.values(), axis=0).sample(frac=1, random_state=7)
```
